### 📦 Importing Libraries

```python
import pandas as pd
```

* **`pandas`** is imported and aliased as `pd`. This library is essential for handling and manipulating structured data, especially in tabular form (DataFrames).

```python
import matplotlib.pyplot as plt
```

* **`matplotlib.pyplot`** is imported as `plt`. This module provides a MATLAB-like plotting framework for visualizing data.

```python
import seaborn as sns
```

* **`seaborn`** is imported as `sns`. It's built on top of `matplotlib` and offers a high-level interface for drawing attractive and informative statistical graphics.

---

### 📥 Reading the Data

```python
df = pd.read_csv('household_power_consumption.txt', \
                 sep=';', \
                 parse_dates={'datetime': ['Date', 'Time']}, \
                 infer_datetime_format=True, \
                 na_values='?', \
                 low_memory=False)
```

Each part of this `read_csv()` function does the following:

| Parameter                                    | Explanation                                                                   |
| -------------------------------------------- | ----------------------------------------------------------------------------- |
| `'household_power_consumption.txt'`          | The path to the dataset file to be read.                                      |
| `sep=';'`                                    | The data uses a semicolon `;` as a separator instead of a comma.              |
| `parse_dates={'datetime': ['Date', 'Time']}` | Merges `Date` and `Time` columns into a single `datetime` column.             |
| `infer_datetime_format=True`                 | Helps speed up parsing by inferring the datetime format automatically.        |
| `na_values='?'`                              | Treats `?` as missing values (`NaN`) in the dataset.                          |
| `low_memory=False`                           | Ensures more accurate data type inference at the cost of higher memory usage. |

### 🧾 Inspecting the Data

```python
df.info()
```

* Displays column names, data types, and counts of non-null values for each column — useful for understanding dataset structure and identifying missing data.

```python
df.describe()
```

* Generates descriptive statistics such as **mean, min, max, and quartiles** for all numeric columns, helping you understand the range and distribution of values.

### 🔍 Handling Missing Data

```python
df.isna().sum()
```

* Checks how many missing (`NaN`) values are present in each column, so you can decide whether to clean or impute them.

```python
df_clean = df.dropna()
```

* Drops any row in the dataset that contains missing values. The cleaned DataFrame is stored as `df_clean`.

---

### 📊 Histogram of Global Active Power

```python
plt.figure(figsize=(10, 4))
```

* Sets up a new figure for plotting with a custom size (10 inches wide and 4 inches high).

```python
df_clean['Global_active_power'].hist(bins=100)
```

* Plots a histogram of the `Global_active_power` variable with **100 bins**, showing its distribution (frequent values, spread, skewness, etc.).

```python
plt.title('Global Active Power Distribution')
```

* Adds a descriptive title to the histogram.

```python
plt.xlabel('Global Active Power (kilowatts)')
```

* Labels the x-axis to show units of measurement.

```python
plt.ylabel('Frequency')
```

* Labels the y-axis to show how often each value range occurs.

```python
plt.show()
```

* Displays the histogram plot.

---

### 📈 Time-Series Plot of Global Active Power

```python
plt.figure(figsize=(14, 4))
```

* Prepares a larger canvas (14x4 inches) for a time-series line plot.

```python
df_clean.set_index('datetime')['Global_active_power'].plot()
```

* First sets `'datetime'` as the index so time is on the x-axis.
* Then plots the `'Global_active_power'` over time to visualize trends, spikes, or patterns.

```python
plt.title('Global Active Power Over Time')
```

* Adds a clear title to the time-series plot.

```python
plt.ylabel('Kilowatts')
```

* Labels the y-axis to indicate power consumption.

```python
plt.xlabel('Time')
```

* Labels the x-axis to indicate time progression.

```python
plt.tight_layout()
```

* Adjusts subplot parameters to give enough padding and avoid label cutoff.

```python
plt.show()
```

* Renders the final time-series plot.
