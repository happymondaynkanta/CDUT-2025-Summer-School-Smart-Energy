
# 🤖 Baseline Modeling with Linear Regression and Random Forest

This notebook provides a **comprehensive, line-by-line explanation** of a baseline regression modeling pipeline using **Linear Regression** and **Random Forest** on household power consumption data.

---

### 📦 Importing Required Libraries

```python
import pandas as pd
```
* Loads the **pandas** library, used for reading and manipulating tabular data (DataFrames).

```python
import numpy as np
```
* Loads **NumPy**, a library for numerical computations.

```python
import matplotlib.pyplot as plt
```
* Imports **matplotlib.pyplot**, which is used for creating visual plots.

```python
import seaborn as sns
```
* Imports **seaborn**, a high-level visualization library built on top of matplotlib.

---

### 🔀 Splitting Data and Model Imports

```python
from sklearn.model_selection import train_test_split
```
* Imports the `train_test_split()` function to split data into training and testing sets.

```python
from sklearn.linear_model import LinearRegression
```
* Imports **Linear Regression**, a basic model that assumes a linear relationship between inputs and target.

```python
from sklearn.ensemble import RandomForestRegressor
```
* Imports **Random Forest**, an ensemble method based on decision trees for robust prediction.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```
* Imports evaluation metrics:
  - **MAE** (Mean Absolute Error)
  - **RMSE** (Root Mean Squared Error)
  - **R² Score** (Explained variance)

---

### 📥 Load and Prepare Dataset

```python
df = pd.read_csv('data/features.csv')
```
* Loads the dataset containing feature-engineered values from a CSV file.

```python
X = df.drop(columns=['Global_active_power', 'datetime'])
```
* Defines input features (`X`) by dropping the target and timestamp columns.

```python
y = df['Global_active_power']
```
* Defines the output target variable (`y`) we want to predict.

---

### 🔀 Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```
* Splits data into 80% training and 20% testing sets.
* `shuffle=False` preserves the temporal order for time-series consistency.

---

### 📈 Linear Regression Model

```python
ols = LinearRegression()
```
* Initializes a **Linear Regression** model instance.

```python
ols.fit(X_train, y_train)
```
* Trains (fits) the model on the training data.

```python
y_pred_ols = ols.predict(X_test)
```
* Uses the trained model to make predictions on the test set.

---

### 🌲 Random Forest Model

```python
rf = RandomForestRegressor(n_estimators=200, random_state=42)
```
* Initializes a **Random Forest** model with 200 trees. `random_state=42` ensures reproducibility.

```python
rf.fit(X_train, y_train)
```
* Trains the Random Forest model on the training data.

```python
y_pred_rf = rf.predict(X_test)
```
* Makes predictions using the trained Random Forest model.

---

### 📊 Model Evaluation Function

```python
def print_metrics(y_true, y_pred, model_name):
    print(f'--- {model_name} ---')
    print(f'MAE: {mean_absolute_error(y_true, y_pred):.3f}')
    print(f'RMSE: {mean_squared_error(y_true, y_pred, squared=False):.3f}')
    print(f'R2: {r2_score(y_true, y_pred):.3f}\n')
```
* Defines a function to print:
  - **MAE**: average magnitude of prediction errors.
  - **RMSE**: square root of average squared errors.
  - **R²**: proportion of variance explained by the model.

---

### 🧪 Print Evaluation Metrics

```python
print_metrics(y_test, y_pred_ols, 'OLS')
```
* Evaluates and prints metrics for **Linear Regression**.

```python
print_metrics(y_test, y_pred_rf, 'Random Forest')
```
* Evaluates and prints metrics for **Random Forest**.

---

### 🎯 Actual vs Predicted Scatter Plot

```python
plt.figure(figsize=(6, 6))
```
* Sets the plot size to 6x6 inches.

```python
plt.scatter(y_test, y_pred_rf, alpha=0.3)
```
* Plots actual vs predicted values as a scatterplot with transparency.

```python
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
```
* Adds a red dashed reference line indicating perfect prediction (y = x).

```python
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Random Forest: Actual vs Predicted')
plt.show()
```
* Adds labels, title, and shows the plot.

---

### 📉 Residual Plot

```python
residuals = y_test - y_pred_rf
```
* Calculates residuals (prediction errors) for the test set.

```python
plt.figure(figsize=(10, 4))
plt.plot(residuals.values)
plt.axhline(y=0, color='r', linestyle='--')
plt.title('Residual Plot')
plt.xlabel('Time')
plt.ylabel('Residual')
plt.show()
```
* Plots the residuals over time to visualize bias and variability.

---

### 🧠 Feature Importance (Random Forest)

```python
importances = rf.feature_importances_
```
* Extracts feature importance values from the trained Random Forest model.

```python
feat_df = pd.DataFrame({'feature': X.columns, 'importance': importances})
feat_df.sort_values(by='importance', ascending=False, inplace=True)
```
* Creates a DataFrame to rank features by their contribution to prediction.

```python
plt.figure(figsize=(10, 5))
sns.barplot(x='importance', y='feature', data=feat_df)
plt.title('Feature Importance')
plt.show()
```
* Visualizes the importance of each feature using a horizontal bar chart.
