
# ⚙️ Feature Engineering for Household Power Data

This notebook provides a **line-by-line explanation** of how to engineer features from a cleaned household power consumption dataset using **pandas**, **NumPy**, and **scikit-learn**.

---

### 📦 Importing Required Libraries

```python
import pandas as pd
```
* **`pandas`** is used for loading and manipulating structured data.

---

### 📥 Load Dataset

```python
df = pd.read_csv('energy_data_cleaned.csv', sep=';')
```
* Loads the cleaned dataset from a CSV file.
* Parses the `'datetime'` column as a `datetime` object for easier time-based operations.

---

### 🔢 Define Numeric Columns

```python
numeric_cols = ['Global_active_power','Global_reactive_power','Voltage',
                'Global_intensity','Sub_metering_1','Sub_metering_2','Sub_metering_3']
```
* Specifies the list of numerical features to be converted and used for further processing.

---

### 🔄 Convert Data Types

```python
df[numeric_cols] = df[numeric_cols].astype(float)
```
* Converts the selected numeric columns to `float` type to ensure compatibility with mathematical operations.

### 🔄 Convert 'datetime' column to datetime format if not already
```python
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
```
---

### 🕒 Extract Date-Time Features

```python
df['hour'] = df['datetime'].dt.hour
```
* Extracts the hour from the timestamp to capture hourly variations in power usage.

```python
df['day_of_week'] = df['datetime'].dt.dayofweek
```
* Extracts the day of the week (0 = Monday, 6 = Sunday) to capture weekly patterns.

---

### 📐 Import NumPy for Feature Encoding

```python
import numpy as np
```
* Imports **NumPy**, used for creating cyclic time-based features (sinusoidal encoding).

---

### 🔁 Cyclic Feature Encoding

```python
df['hour_sin'] = np.sin(2 * np.pi * df['hour']/24)
```
* Encodes hour of the day as a **sine wave** to model the cyclical nature of time.

```python
df['hour_cos'] = np.cos(2 * np.pi * df['hour']/24)
```
* Complements `hour_sin` to provide a two-dimensional representation of cyclical hour values.

---

### 🧪 Feature Scaling

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```
* Imports two types of scalers:
  - **`StandardScaler`**: Scales features to have zero mean and unit variance.
  - **`MinMaxScaler`**: Scales features to a fixed range [0, 1].

```python
scaler_std = StandardScaler()
```
* Initializes standard scaler.

```python
scaler_minmax = MinMaxScaler()
```
* Initializes min-max scaler.

---

### 🔃 Transform Numeric Data

```python
df_std = pd.DataFrame(scaler_std.fit_transform(df[numeric_cols]), columns=numeric_cols)
```
* Applies standard scaling to numeric columns and returns a new DataFrame.

```python
df_minmax = pd.DataFrame(scaler_minmax.fit_transform(df[numeric_cols]), columns=numeric_cols)
```
* Applies min-max scaling to numeric columns and returns another scaled DataFrame.

---

### ⚡ Derived Feature: Apparent Power

```python
df['Apparent_Power'] = (df['Global_active_power']**2 + df['Global_reactive_power']**2)**0.5
```
* Computes **Apparent Power** using the formula:  
  $$
  \text{Apparent Power} = \sqrt{(\text{Active Power})^2 + (\text{Reactive Power})^2}
  $$
* This gives a more complete view of energy usage in the household.

---

### 💾 Export the Feature Set

```python
df.to_csv('data/features.csv', index=False)
```
* Saves the enhanced dataset with new features to a new CSV file for later use in modeling or analysis.
